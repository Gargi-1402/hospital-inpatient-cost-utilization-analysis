```python
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

use_cols = [
    'Hospital Service Area',
    'Hospital County',
    'Permanent Facility Id',
    'Facility Name',
    'Age Group',
    'Zip Code - 3 digits',
    'Gender',
    'Race',
    'Ethnicity',
    'Length of Stay',
    'Type of Admission',
    'Patient Disposition',
    'Discharge Year',
    'CCSR Diagnosis Code',
    'CCSR Diagnosis Description',
    'CCSR Procedure Code',
    'CCSR Procedure Description',
    'APR DRG Code',
    'APR DRG Description',
    'APR MDC Code',
    'APR MDC Description',
    'APR Severity of Illness Code',
    'APR Severity of Illness Description',
    'APR Risk of Mortality',
    'APR Medical Surgical Description',
    'Payment Typology 1',
    'Emergency Department Indicator',
    'Total Charges',
    'Total Costs'
]

output_file = "/content/drive/MyDrive/SPARCS Project/SPARCS_2023_clean.parquet"

writer = None
total_processed = 0

for chunk in pd.read_csv(
    file_path,
    usecols=use_cols,
    chunksize=100_000,
    low_memory=False
):

    # Clean Length of Stay
    chunk['Length of Stay'] = pd.to_numeric(
        chunk['Length of Stay'],
        errors='coerce'
    )

    # Clean financial columns
    for col in ['Total Charges', 'Total Costs']:
        chunk[col] = pd.to_numeric(
            chunk[col].astype(str).str.replace(',', '', regex=False),
            errors='coerce'
        )

    # Convert chunk to Arrow table
    table = pa.Table.from_pandas(chunk, preserve_index=False)

    # Create writer using first chunk's schema
    if writer is None:
        writer = pq.ParquetWriter(
            output_file,
            table.schema,
            compression='snappy'
        )

    writer.write_table(table)

    total_processed += len(chunk)
    print(f"Processed: {total_processed:,} rows")

# Close writer
if writer is not None:
    writer.close()

print("\nDONE")
print(f"Total rows processed: {total_processed:,}")
print(f"Saved to: {output_file}")
```

```python
cost_by_severity = (
    df.groupby('APR Severity of Illness Description')
     .agg(
          Patients=('Total Costs', 'size'),
          Median_Cost=('Total Costs', 'median'),
          Mean_Cost=('Total Costs', 'mean'),
          Median_LOS=('Length of Stay', 'median'),
          Mean_LOS=('Length of Stay', 'mean')
      )
      .sort_values('Median_Cost')
)

cost_by_severity
```

```python
cost_by_admission = (
    df.groupby('Type of Admission')
      .agg(
          Patients=('Total Costs', 'size'),
          Median_Cost=('Total Costs', 'median'),
          Mean_Cost=('Total Costs', 'mean'),
          Median_LOS=('Length of Stay', 'median'),
          Mean_LOS=('Length of Stay', 'mean')
      )
      .sort_values('Median_Cost', ascending=False)
)

cost_by_admission
```

```python
admission_analysis = (
    df.groupby('Type of Admission')
      .agg(
          Admissions=('Total Costs', 'size'),
          Total_Cost=('Total Costs', 'sum'),
          Median_Cost=('Total Costs', 'median'),
          Mean_Cost=('Total Costs', 'mean')
      )
)

admission_analysis['Cost_Share_%'] = (
    admission_analysis['Total_Cost']
    / admission_analysis['Total_Cost'].sum()
    * 100
)

admission_analysis['Volume_Share_%'] = (
    admission_analysis['Admissions']
    / admission_analysis['Admissions'].sum()
    * 100
)

admission_analysis.sort_values('Total_Cost', ascending=False)
```

```python
diagnosis_analysis = (
    df.groupby(['CCSR Diagnosis Code', 'CCSR Diagnosis Description'])
      .agg(
          Admissions=('Total Costs', 'size'),
          Total_Cost=('Total Costs', 'sum'),
          Median_Cost=('Total Costs', 'median'),
          Median_LOS=('Length of Stay', 'median')
      )
      .reset_index()
)

diagnosis_analysis['Cost_Share_%'] = (
    diagnosis_analysis['Total_Cost']
    / diagnosis_analysis['Total_Cost'].sum()
    * 100
)

diagnosis_analysis = diagnosis_analysis.sort_values(
    'Total_Cost', ascending=False
)

diagnosis_analysis.head(20)
```

```python
procedure_analysis = (
    df.dropna(subset=['CCSR Procedure Code'])
      .groupby(['CCSR Procedure Code', 'CCSR Procedure Description'])
      .agg(
          Admissions=('Total Costs', 'size'),
          Total_Cost=('Total Costs', 'sum'),
          Median_Cost=('Total Costs', 'median'),
          Median_LOS=('Length of Stay', 'median')
      )
      .reset_index()
)

procedure_analysis['Cost_Share_%'] = (
    procedure_analysis['Total_Cost']
    / df['Total Costs'].sum()
    * 100
)

procedure_analysis = procedure_analysis.sort_values(
    'Total_Cost', ascending=False
)

procedure_analysis.head(20)
```

```python
df['Cost_per_Day'] = (
    df['Total Costs'] / df['Length of Stay']
)

df['Cost_per_Day'].describe()
```

```python
cost_intensity = (
    df.groupby('APR Severity of Illness Description')
      .agg(
          Admissions=('Total Costs', 'size'),
          Median_Total_Cost=('Total Costs', 'median'),
          Median_LOS=('Length of Stay', 'median'),
          Median_Cost_per_Day=('Cost_per_Day', 'median')
      )
      .sort_values('Median_Cost_per_Day', ascending=False)
)

cost_intensity
```

```python
high_cost_threshold = df['Total Costs'].quantile(0.95)

print(f"95th percentile cost: ${high_cost_threshold:,.2f}")

df['High_Cost'] = df['Total Costs'] >= high_cost_threshold

print("\nHigh-cost admissions:")
print(df['High_Cost'].value_counts())

print("\nHigh-cost share:")
print(df['High_Cost'].mean() * 100)
```

```python
high_cost_profile = (
    df.groupby('High_Cost')
      .agg(
          Admissions=('Total Costs', 'size'),
          Median_Cost=('Total Costs', 'median'),
          Median_LOS=('Length of Stay', 'median'),
          Mean_Cost=('Total Costs', 'mean')
      )
)

high_cost_profile
```

```python
high_cost_severity = (
    df.groupby('APR Severity of Illness Description')
      .agg(
          Total_Admissions=('Total Costs', 'size'),
          High_Cost_Admissions=('High_Cost', 'sum')
      )
)

high_cost_severity['High_Cost_Rate_%'] = (
    high_cost_severity['High_Cost_Admissions']
    / high_cost_severity['Total_Admissions']
    * 100
)

high_cost_severity.sort_values(
    'High_Cost_Rate_%',
    ascending=False
)
```

```python
high_cost_diagnosis = (
    df.groupby(
        ['CCSR Diagnosis Code', 'CCSR Diagnosis Description']
    )
    .agg(
        Admissions=('High_Cost', 'size'),
        High_Cost_Admissions=('High_Cost', 'sum')
    )
    .reset_index()
)

high_cost_diagnosis['High_Cost_Rate_%'] = (
    high_cost_diagnosis['High_Cost_Admissions']
    / high_cost_diagnosis['Admissions']
    * 100
)

high_cost_diagnosis = high_cost_diagnosis[
    high_cost_diagnosis['Admissions'] >= 1000
].sort_values(
    'High_Cost_Rate_%',
    ascending=False
)

high_cost_diagnosis.head(20)
```

```python
hospital_analysis = (
    df.groupby(['Permanent Facility Id', 'Facility Name'])
      .agg(
          Admissions=('Total Costs', 'size'),
          Median_Cost=('Total Costs', 'median'),
          Mean_Cost=('Total Costs', 'mean'),
          Median_LOS=('Length of Stay', 'median'),
          Total_Cost=('Total Costs', 'sum')
      )
      .reset_index()
)

hospital_analysis = hospital_analysis[
    hospital_analysis['Admissions'] >= 1000
].sort_values(
    'Median_Cost',
    ascending=False
)

hospital_analysis.head(20)
```
